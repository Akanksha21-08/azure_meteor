```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('ChurnDataset.csv') #read the dataset
df #to show the dataset
```

```python
df.head()#to show first five rows
```

```python
df.info() #inspection of data
```

#replacing the blanks with 0 as tenure is 0 and no total charges are recorded

```python
df["TotalCharges"]=df["TotalCharges"].replace(" ","0")
df["TotalCharges"]=df["TotalCharges"].astype("float")
```

```python
df.info()
```

```python
df.isnull().sum() #isnull() shows the null values it returns true if there is any null value
```

```python
df.describe() #descriptive analysis
```

```python
df.duplicated() #for duplicate values it will return true
```

```python
df.duplicated().sum()
```

```python
df["customerID"].duplicated().sum()
```

```python
def conv(value):
    if value==1:
        return "yes"
    else:
        return "no"

df['SeniorCitizen']=df["SeniorCitizen"].apply(conv)
        
```

#converted 0 and 1 values of senior citizen to yes/no to make it easier to understand

```python
df.head(50)
```

```python
ax=sns.countplot(x='Churn',data=df)
ax.bar_label(ax.containers[0])#this line will show the no. of churned out customers
plt.title("Count of customers by churn")
plt.show()
```

```python
gb=df.groupby("Churn").agg({"Churn":"count"})#we are grouping the string data type and counting it because for pie chart float value is needed
plt.pie(gb['Churn'])
plt.title("Percentage of Churned Customers")
plt.show()
gb
```

```python
plt.pie(gb['Churn'],labels=gb.index,autopct="%1.2f%%")#this line will show the yes/no labels and percentange also
plt.title("Percentage of Churned Customers")#used to show title in plots
plt.show
```

#from the given pie chart we can conclude that 26.54% of our customers have churned out
#now let's explore the reason behind it

```python
sns.countplot(x="gender",data=df)#count of male and female customers
plt.show()
```

```python
sns.countplot(x="gender",data=df,hue="Churn")#count of male and female  churned out customers
plt.title("Churn by Gender")
plt.show()
```

```python
ax=sns.countplot(x="SeniorCitizen",data=df)
ax.bar_label(ax.containers[0])
plt.title("Count of Customers by Senior Citizen")
plt.show()
```

```python
sns.countplot(x="SeniorCitizen",data=df,hue="Churn")#count of churned out senior citizens 
plt.title("Churn by Senior Citizen")
plt.show()
```

```python
ax = sns.countplot(x="SeniorCitizen", data=df, hue="Churn")

# Add count labels
for container in ax.containers:
    ax.bar_label(container)

plt.xticks([0, 1], ["No", "Yes"])
plt.title("Churn by Senior Citizen(Total Values)")
plt.show()
```

#comparitively a greater percentage of people in senior citizen have churned

```python
plt.figure(figsize=(9,4))
sns.histplot(x="tenure",data=df,bins=72,hue="Churn")#bins-distribution
plt.show()
```

#people who have used our services for long time have stayed and people who have used our services for 1 or 2 months have churned

```python
ax=sns.countplot(x="Contract",data=df,hue="Churn")#count of churned customers according to contract
ax.bar_label(ax.containers[0])
plt.title("Count of Customers by Contract")
plt.show()
```

#people who have month to month contact likely to churnthen for those who have 1 or 2 years of contract

```python
cols = [
    'PhoneService', 'MultipleLines', 'InternetService',
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
    'TechSupport', 'StreamingTV', 'StreamingMovies'
]

# Create subplot grid (3 rows × 3 columns)
fig, axes = plt.subplots(3, 3, figsize=(15, 12))

for i, col in enumerate(cols):
    row = i // 3
    col_idx = i % 3
    
    sns.countplot(x=col, data=df, ax=axes[row, col_idx])
    axes[row, col_idx].set_title(col)

plt.tight_layout()
plt.show()
```

#subplot for better analysis on the basis of services

```python
cols = [
    'PhoneService', 'MultipleLines', 'InternetService',
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
    'TechSupport', 'StreamingTV', 'StreamingMovies'
]

# Create subplot grid (3 rows × 3 columns)
fig, axes = plt.subplots(3, 3, figsize=(15, 12))

for i, col in enumerate(cols):
    row = i // 3
    col_idx = i % 3
    
    sns.countplot(x=col, data=df, ax=axes[row, col_idx],hue=df["Churn"])
    axes[row, col_idx].set_title(col)

plt.tight_layout()
plt.show()
```

#Customers without services like OnlineSecurity, TechSupport, and Backup show higher churn, indicating these features help retain users.
#Fiber optic internet users have higher churn compared to DSL and no internet service.
#Users with PhoneService and MultipleLines show moderate churn.
#ustomers with No internet service generally have the lowest churn.
